## Define your Source Directory and Target Directory
both the paths can be taken/copied from inside the Databricks, under the catalog where you created the volume
- source_dir = the volume path to your source directory
- target_dir = the volume path of your archive directory 

```python
source_dir = "/Volumes/dev_ext_loc/default/dbx-event-driven-source/source/"
target_dir = "/Volumes/dev_ext_loc/default/dbx-event-driven-source/archive/"
```

### Create a Staging Table 
create a staging table under the catalog where you created volume fir external location

```python
stage_table = "dev_ext_loc.default.stg_table"
```

## Read the data from the Source Directory

```python
df = spark.read.csv(source_dir, headrer=True, inferSchema=True)
```

### Write the data in  Staging Table
overwrite the data in the staging table

```python
df.write.format("delta").mode("overwrite").saveAsTable(stage_table)
```

## Move the file in the Archive Directory
loop over all the files and move all of them in archive 

```python
# list all the files
files = dbutils.fs.ls(source_dir)

# iterate on the list and print each file path 
for file in files:
    src_path = file.path

    # construct the target path
    target_path = target_dir + src_path.split("/")[-1]

    # move the file
    dbutils.fs.mv(src_path, target_path)
```
